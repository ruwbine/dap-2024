### **2.1 Типы ячеек в Jupyter Notebook**
В Jupyter Notebook используются три типа ячеек:
1. **Code** — выполнение программного кода.  
2. **Markdown** — оформление текста с разметкой.  
3. **Raw** — необработанный текст без форматирования.

### **2.2 Как с помощью Markdown сформировать иерархический список?**

Используются символы `-`, `*` или `+` и отступы:

```

* Первый уровень

  * Второй уровень

    * Третий уровень

```

### **2.3 Как оформить цитату в Markdown?**

Используется символ `>` в начале строки:

```

> Это цитата.

`````


### **2.4 Как вставить отформатированный блок кода в текстовую ячейку?**

Используются тройные обратные кавычки:

```python
print("Hello, world!")
```

### **2.5 Сколько уровней заголовков существует в Markdown?**

В Markdown существует **6 уровней заголовков**:

```
# Заголовок 1
## Заголовок 2
### Заголовок 3
#### Заголовок 4
##### Заголовок 5
###### Заголовок 6
```

### 2.6 Шаги процесса анализа данных и подробное описание одного шага

Основные шаги анализа данных:
1. Постановка задачи  
2. Сбор данных  
3. Очистка данных  
4. Разведочный анализ (EDA)  
5. Подготовка данных  
6. Построение модели  
7. Оценка результатов  
8. Интерпретация и визуализация  

Очистка данных — это процесс удаления пропусков, исправления ошибок, фильтрации дубликатов и аномалий.  
На этом этапе:
- заполняются или удаляются пропущенные значения,
- исправляются форматы дат и чисел,
- удаляются дубликаты,
- устраняются выбросы и ошибочные данные.

Качество очистки напрямую влияет на точность анализа и моделей.

### **2.7 Типы данных и описание одного из них**

Основные типы данных в Python:
- `int`
- `float`
- `str`
- `bool`
- `list`
- `tuple`
- `dict`
- `set`

**Подробно: dict (словарь)**  
Словарь — структура данных вида *ключ : значение*.  
Особенности:
- доступ по ключу очень быстрый,
- ключи уникальны,
- удобен для хранения структурированных данных.

Пример:
```python
user = {"name": "Nikita", "age": 20}
```

### **2.8 Что такое Jupyter Notebook и для чего используется?**

Jupyter Notebook — это интерактивная среда, позволяющая запускать код и добавлять текстовые описания.  
Используется для:
- анализа данных,
- машинного обучения,
- визуализации,
- прототипирования,
- обучения.


### **2.9 Почему Python популярен для анализа данных?**

- Прост и читаем.  
- Имеет мощные библиотеки: `pandas`, `numpy`, `matplotlib`, `scikit-learn`.  
- Большое сообщество.  
- Отлично интегрируется с Jupyter.  
- Универсален: от ML до backend.


### **2.10 Обязательно ли использовать Jupyter Notebook для анализа данных?**

Нет, не обязательно.  
Можно использовать:
- Python-скрипты,
- PyCharm/VSCode,
- Google Colab,
- RStudio,
- Excel/Power BI.

Но Jupyter удобен из-за интерактивности и визуализации.
