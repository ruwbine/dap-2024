Задача №1. https://leetcode.com/problems/two-sum/

````python
class Solution1(object):
    def twoSum(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: List[int]
        """
        # Ваше решение
        remain = {}

        for i, num in enumerate(nums):
            complement = target - num
            if complement in remain:
                return [i, remain.get(complement)]
            remain[num] = i
        return []
````

Задача №2. https://leetcode.com/problems/palindrome-number/description/

````python
class Solution2(object):
    def isPalindrome(self, x):
        """
        :type x: int
        :rtype: bool
        """
        # Ваше решение
        str_x = str(x)
        if x < 0:
            return False
        return str_x == str_x[::-1]
````

Задача №3. https://leetcode.com/problems/roman-to-integer/description/

````python
class Solution3(object):
    def romanToInt(self, s):
        """
        :type s: str
        :rtype: int
        """

        # Ваше решение
        romains = {
            'I': 1,
            'V': 5,
            'X': 10,
            'L': 50,
            'C': 100,
            'D': 500,
            'M': 1000
        }

        total_sum = 0

        for i in range(len(s) - 1):
            if romains[s[i]] < romains[s[i+1]]:
                total_sum -= romains[s[i]]
            else:
                total_sum += romains[s[i]]
        return total_sum + romains[s[-1]]
````

Задача №5. https://leetcode.com/problems/merge-two-sorted-lists/description/

````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


class Solution5(object):
    def mergeTwoLists(self, list1, list2):
        """
        :type list1: Optional[ListNode]
        :type list2: Optional[ListNode]
        :rtype: Optional[ListNode]
        """

        # Ваше решение
        dummyNode = ListNode(-1)
        temp = dummyNode

        curr1, curr2 = list1, list2

        while curr1 and curr2:
            if curr1.val < curr2.val:
                temp.next = curr1
                curr1 = curr1.next
            else:
                temp.next = curr2
                curr2 = curr2.next
            temp = temp.next

        temp.next = curr1 if curr1 else curr2

        return dummyNode.next
````

### **1. Что такое динамическая типизация и чем она отличается от статической?**

**Динамическая типизация** — тип переменной определяется во время выполнения программы, а не во время компиляции.
В Python можно написать:

```python
x = 10
x = "text"  # тип меняется динамически
```

**Статическая типизация** (C++, Java):
Тип переменной задаётся при объявлении и не может измениться:

```cpp
int x = 10;
x = "text"; // ошибка
```

**Разница:**

* динамическая — гибче, проще писать, но ошибки находятся на этапе выполнения;
* статическая — безопаснее и быстрее, но строже.

### **2. Чем отличается цикл `for` в Python от цикла `for` в C++?**

**В Python** цикл `for` — *итератор по коллекции*, а не счётчик:

```python
for x in [1, 2, 3]:
    print(x)
```

Он проходит по элементам списка, строки, генератора и т.д.

**В C++** классический `for` — это *цикл со счётчиком*:

```cpp
for (int i = 0; i < 3; i++) {
    cout << i;
}
```

В C++ можно тоже итерировать коллекции, но синтаксис другой.

### **3. Для чего предназначены операторы `continue` и `break`?**

* **break** — полностью завершает цикл.
* **continue** — пропускает текущую итерацию и переходит к следующей.

Пример:

```python
for x in range(5):
    if x == 2:
        continue  # пропустить 2
    if x == 4:
        break     # остановить цикл
    print(x)
```

### **4. Как определить функцию в Python? Может ли функция возвращать несколько значений?**

Функция определяется с помощью ключевого слова `def`:

```python
def add(a, b):
    return a + b
```

**Функция может возвращать несколько значений**, но фактически Python возвращает **кортеж**:

```python
def f():
    return 1, 2, 3

x, y, z = f()
```

### **5. Что такое лямбда-функции и для чего они предназначены?**

**Лямбда-функции** — это маленькие анонимные функции, которые записываются компактно:

```python
square = lambda x: x * x
```

Используются для:

* сортировки (`key`),
* функционального программирования (`map`, `filter`),
* однострочных вычислений.

### **6. Может ли тело функции содержать определение другой функции?**

Да. Python поддерживает **вложенные функции**:

```python
def outer():
    def inner():
        print("Внутренняя функция")
```

Это используется, например, в замыканиях и декораторах.


### **7. Как определить класс в Python? Для чего нужна переменная `self`?**

Определение класса:

```python
class Person:
    def __init__(self, name):
        self.name = name
```

**self** — ссылка на текущий объект (аналог `this` в C++/Java).
Через `self` обращаются к полям и методам объекта.


### **8. Как определить приватный метод? Можно ли вызвать приватный метод за пределами класса?**

Приватные методы определяются с помощью двойного подчёркивания:

```python
class A:
    def __hidden(self):
        print("Приватный метод")
```

Python выполняет *name mangling* → имя превращается в `_A__hidden`.

**Вызвать можно**, но только так:

```python
obj = A()
obj._A__hidden()  # технически возможно, но не рекомендуется
```

### **9. Как определить статический метод класса? Для чего может понадобиться статический метод?**

Используется декоратор `@staticmethod`:

```python
class Math:
    @staticmethod
    def add(a, b):
        return a + b
```

**Статический метод:**

* не получает параметров `self` или `cls`,
* не зависит от состояния объекта или класса.

Используется как обычная полезная функция, но логически связанная с классом.

### **10. Чем `staticmethod` отличается от `classmethod`?**

| Метод          | Что получает первым аргументом | Использование                      |
| -------------- | ------------------------------ | ---------------------------------- |
| `staticmethod` | ничего                         | обычная функция внутри класса      |
| `classmethod`  | **класс (`cls`)**              | работа с классом, фабричные методы |

Пример:

```python
class A:
    @classmethod
    def create_default(cls):
        return cls(10)

    @staticmethod
    def util(x):
        return x * 2
```

### **11. Как определить собственный класс исключений Python? Как выбросить и перехватить исключение?**

Создать класс, унаследованный от `Exception`:

```python
class MyError(Exception):
    pass
```

Выбросить:

```python
raise MyError("Ошибка!")
```

Перехватить:

```python
try:
    raise MyError("Проблема")
except MyError as e:
    print(e)
```

### **12. Что такое кортеж, для чего он используется?**

**Кортеж** (`tuple`) — неизменяемая последовательность элементов:

```python
t = (1, 2, 3)
```

Используется для:

* хранения набора значений, которые нельзя изменить,
* возврата нескольких значений из функции,
* ключей словаря (если элементы тоже неизменяемы),
* безопасной передачи данных.
