6. Дан массив 10x2 подеставлющий собой точки в декартовой системе координат, преобразовать в полярную

````python
z = np.random.random((10,2))
# Ваш код здесь
x, y = z[:,0], z[:,1]
r = np.sqrt(x**2 + y**2)
phi = np.arctan2(y, x)
polar = np.column_stack([r, phi])
print(polar)
````

**10**.  В чем преимущество массивов NumPy по сравнению с обычными списками.  Чем обусловлено это преимущество?

Основное преимущество: скорость за счёт векторизации и работы в C

NumPy хранит данные в непрерывном блоке памяти и использует оптимизированный код на C/Fortran, что делает операции в десятки раз быстрее, чем Python-списки.
