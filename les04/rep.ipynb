## **1. Что такое медиана?**

Медиана — это число, которое находится в середине упорядоченного ряда.
То есть половина значений меньше него, а половина — больше.

## **2. Что такое дисперсия?**

Дисперсия показывает, насколько сильно значения данных отклоняются от среднего значения.
Если дисперсия большая — данные сильно разбросаны.
Если маленькая — данные близки друг к другу.

## **3. Что такое среднеквадратичное отклонение?**

Среднеквадратичное отклонение — это показатель, который показывает «типичное» отклонение значений от среднего.
Если оно маленькое — данные почти одинаковые.
Если большое — значения сильно отличаются друг от друга.

## **4. Что такое Series и чем он отличается от DataFrame?**

**Series** — это одномерный объект, похожий на список, но с индексами.
**DataFrame** — это таблица, в которой много столбцов, и каждый столбец — это Series.

Разница:

* Series — одна колонка
* DataFrame — много колонок, полноценная таблица


## **5. Можно ли изменить индекс в Series после создания объекта?**

Да, можно изменить индекс в любой момент


## **6. Можно ли выполнять сложение двух Series в разных ситуациях?**

pandas всегда пытается выполнить сложение. Разница — в результате:

### **Одинаковые типы, одинаковые размеры, одинаковые индексы**

Да, сложение произойдёт поэлементно.

### **Одинаковые типы, одинаковые размеры, разные индексы**

Да, но значения будут складываться только там, где индексы совпадают.
Несовпадающие индексы дадут пропуски.

### **Одинаковые типы, разные размеры, одинаковые индексы**

Да.
Для отсутствующих значений появятся пропуски (NaN).

### **Разные типы данных, одинаковые размеры и одинаковые индексы**

Да.
Типы будут автоматически приведены (например, числа станут дробными).

### **Разные типы данных, одинаковые размеры и разные индексы**

Да.
Опять же, данные будут выровнены по индексам → где не совпадает, будут пропуски.

### **Разные типы, разные размеры, разные индексы**

Да.
Будет много пропущенных значений из-за несовпадения индексов.

**Итог:** сложение всегда выполняется, но качество результата зависит от совпадения индексов.

## **7. Что такое DataFrame и чем он отличается от Series?**

DataFrame — это двумерная таблица, похожая на Excel: строки × столбцы.
Series — это один столбец.

Главное отличие:

* Series — 1 измерение
* DataFrame — 2 измерения

## **9. Чем отличается iloc от loc?**

* **iloc** — работает по числовым позициям (например: первая строка, вторая строка).
* **loc** — работает по реальным индексам и именам (например: строка с индексом "2024" или "id_5").

## **11. Нужны ли дополнительные библиотеки для загрузки разных форматов?**

* **CSV** — нет, ничего дополнительно ставить не нужно.
* **Excel (.xls, .xlsx)** — да, нужна дополнительная библиотека (например, *openpyxl*).
* **JSON** — нет, загружается без дополнительных установок.
